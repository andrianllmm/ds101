## Chapter 1: Selecting Data

SQL retrieves data using `SELECT` and `FROM`.

```sql
SELECT column_name
FROM table_name;
```

`COUNT` counts rows or non-null values

```sql
SELECT COUNT(*) FROM films;      -- counts all rows
SELECT COUNT(title) FROM films;  -- counts non-null titles
```

`DISTINCT` removes duplicate values

```sql
SELECT DISTINCT language
FROM films;
```

SQL execution order differs from written order

```sql
FROM -> SELECT -> LIMIT
```

SQL style:
- Uppercase keywords
- Indentation
- Aliases for readability

# Chapter 2: Filtering Records

`WHERE` filters rows

```sql
SELECT *
FROM films
WHERE release_year > 2000;  -- 2001 or later
```

Combine conditions with `AND`, `OR`

```sql
SELECT *
FROM films
WHERE release_year > 2000
AND language = 'English';  -- English films from 2001 or later
```

We can also use `BETWEEN` as a shorthand

```sql
SELECT *
FROM films
WHERE release_year BETWEEN 1990 AND 2000;  -- = >= 1990 AND <= 2000
```

Pattern matching

```sql
SELECT *
FROM films
WHERE title LIKE 'A%';  -- starts with A
```

Handle missing values with `NULL`

```sql
SELECT *
FROM films
WHERE budget IS NULL;  -- no budget
```

# Chapter 3: Aggregate Functions

Aggregation summarizes data.

```sql
SELECT
    AVG(rating),  -- Average
    MAX(rating),  -- Maximum
    MIN(rating)   -- Minimum
FROM films;
```

`GROUP BY` splits the data into groups.

```sql
SELECT language, AVG(rating)  -- 2001 or later
FROM films
GROUP BY language;
```

`HAVING` filters the groups.

```sql
SELECT language, AVG(rating) AS avg_rating
FROM films
GROUP BY language
HAVING AVG(rating) > 7;  -- like WHERE
```

# Chapter 4: Sorting and Limiting Results

`ORDER BY` sorts results `ASC` or `DESC`

```sql
SELECT title, rating
FROM films
ORDER BY rating DESC;
```

`LIMIT` restricts the number of results

```sql
SELECT *
FROM films
LIMIT 10;  -- show only the first 10
```

`OFFSET` skips a number of results

```sql
SELECT *
FROM films
LIMIT 10 OFFSET 5; -- show only starting from 6th
```

We can also combine them together

```sql
SELECT title, rating
FROM films
ORDER BY rating DESC
LIMIT 5;
```

# Project: Analyzing Students' Mental Health

Project Link: https://projects.datacamp.com/projects/1593

Task: Explore and analyze the students data to see how the length of stay (stay) impacts the average mental health diagnostic scores of the international students present in the study.

- Return a table with nine rows and five columns.
- The five columns should be aliased as: `stay`, `count_int`, `average_phq`, `average_scs`, and `average_as`, in that order.
- The average columns should contain the average of the `todep` (PHQ-9 test), `tosc` (SCS test), and `toas` (ASISS test) columns for each length of stay, rounded to two decimal places.
- The `count_int` column should be the number of international students for each length of stay.
- Sort the results by the length of stay in descending order.

````python
-- My solution
SELECT
    stay,
    COUNT(*) AS count_int,
    ROUND(AVG(todep), 2) AS average_phq,
    ROUND(AVG(tosc), 2) AS average_scs,
    ROUND(AVG(toas), 2) AS average_as
FROM students
WHERE inter_dom = 'Inter'
    AND stay IS NOT NULL
GROUP BY stay
ORDER BY stay DESC;
````
