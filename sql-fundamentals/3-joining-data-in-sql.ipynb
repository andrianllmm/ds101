## Chapter 1: Inner Joins

An INNER JOIN returns only rows where there is a match in both tables based on a join condition.

```sql
SELECT columns
FROM table1
INNER JOIN table2
ON table1.key = table2.key;
```

Example:

```sql
SELECT 
    customers.customer_id,
    customers.name,
    orders.order_id,
    orders.amount
FROM customers
INNER JOIN orders
ON customers.customer_id = orders.customer_id;
```

We can also use aliases to make the query more readable.

```sql
SELECT 
    c.name,
    o.amount
FROM customers AS c
INNER JOIN orders AS o
ON c.customer_id = o.customer_id;
```

There's also a neat shortcut using the `USING` keyword.

```sql
SELECT 
    name,
    order_id,
    amount
FROM customers
INNER JOIN orders
USING (customer_id);
```

## Chapter 2.1: Outer Joins

Outer joins return matching rows plus non-matching rows from one or both tables.

### Left Outer Join

Returns all records from the left table and matched records from the right.

```sql
SELECT 
    c.name,
    o.order_id
FROM customers c
LEFT JOIN orders o
ON c.customer_id = o.customer_id;
```

### Right Outer Join

Returns all records from the right table and matched records from the left.

```sql
SELECT 
    c.name,
    o.order_id
FROM customers c
RIGHT JOIN orders o
ON c.customer_id = o.customer_id;
```

### Full Outer Join

Returns all records from both tables and matched records from both.

```sql
SELECT 
    c.name,
    o.order_id
FROM customers c
FULL JOIN orders o
ON c.customer_id = o.customer_id;
```

## Chapter 2.2: Cross Joins and Inner Joins

### Cross Joins

A CROSS JOIN returns the Cartesian product of two tables.

```sql
SELECT 
    c.name,
    p.product_name
FROM customers c
CROSS JOIN products p;
```

### Self Joins

A self join joins a table with itself.

```sql
SELECT 
    e1.name AS employee,
    e2.name AS manager
FROM employees e1
INNER JOIN employees e2
ON e1.manager_id = e2.employee_id;
```

## Chapter 3: Set Theory for SQL Joins

### UNION

Combines results and removes duplicates.

```sql
SELECT city FROM customers
UNION
SELECT city FROM suppliers;
```

### UNION ALL

Keeps duplicates.

```sql
SELECT city FROM customers
UNION ALL
SELECT city FROM suppliers;
````

### INTERSECT

Returns common rows between queries.

```sql
SELECT city FROM customers
INTERSECT
SELECT city FROM suppliers;
```

### EXCEPT

Returns rows in first query not in second.

```sql
SELECT city FROM customers
EXCEPT
SELECT city FROM suppliers;
```

## Chapter 4: Subqueries

A query nested inside another query.

### Scalar Subqueries

A subquery that returns a single value.

```sql
SELECT name
FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM orders
    WHERE amount > 1000
);
```

### Subquery in FROM

```sql
SELECT avg_amount
FROM (
    SELECT AVG(amount) AS avg_amount
    FROM orders
) AS avg_table;
```

### Subquery in WHERE

```sql
SELECT name
FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM orders
    WHERE amount > 1000
);
```

### Correlated Subqueries

```
SELECT name
FROM customers c
WHERE (
    SELECT COUNT(*)
    FROM orders o
    WHERE o.customer_id = c.customer_id
) > 3;
```

---

# Mini Application Project: E-commerce Sales Analysis

**Objective**: Analyze customer purchasing behavior using joins, set operations, and subqueries.

**Tables Used**:

customers(customer_id, name, city)
orders(order_id, customer_id, order_date, amount)
products(product_id, product_name)
order_items(order_id, product_id, quantity)

## Customer Order Summary

```sql
SELECT 
    c.name,
    COUNT(o.order_id) AS total_orders,
    SUM(o.amount) AS total_spent
FROM customers c
LEFT JOIN orders o
ON c.customer_id = o.customer_id
GROUP BY c.name;
```

## Top Spenders Above Average Spend

```sql
SELECT name
FROM customers
WHERE customer_id IN (
    SELECT customer_id
    FROM orders
    GROUP BY customer_id
    HAVING SUM(amount) > (
        SELECT AVG(amount) FROM orders
    )
);
```

## Products Purchased Together (Cross Join Insight)

```sql
SELECT
    p1.product_name AS product_a,
    p2.product_name AS product_b
FROM products p1
CROSS JOIN products p2
WHERE p1.product_id < p2.product_id;
```

## Customers in One City but Not Another (Set Logic)

```sql
SELECT city FROM customers
WHERE city IN ('Cebu', 'Manila')

EXCEPT

SELECT city FROM customers
WHERE city IN ('Davao');
```

## Manager-Employee Structure (Self Join)

```sql
SELECT 
    e.name AS employee,
    m.name AS manager
FROM employees e
LEFT JOIN employees m
ON e.manager_id = m.employee_id;
```

---

# Takeways

- INNER JOIN: only matching rows
- OUTER JOIN: includes unmatched data
- CROSS JOIN: all combinations
- SELF JOIN: hierarchical relationships
- SET OPERATIONS: combine query results logically
- SUBQUERIES: enable nested analytical logic
